```python
new_x=np.concatenate((ones,x),axis=1)
new_x
```

```python
res1=np.linalg.inv(np.dot(np.transpose(new_x),new_x))
#print(res1)
res1.shape
```

```python
res2=np.dot(np.transpose(new_x),y)
res2.shape
```

```python
res=np.dot(res1,res2)
print(res)
```
